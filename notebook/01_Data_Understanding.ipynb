```python
fraud_count = fraud_df["Class"].sum()
legitimate_count = (fraud_df["Class"] == 0).sum()

fraud_ratio = legitimate_count / fraud_count

print("Fraud transactions     :", fraud_count)
print("Legitimate transactions:", legitimate_count)
print("Legitimate : Fraud     :", round(fraud_ratio, 2), ": 1")
```
